## Leitura de dados

```python
# bibliotecas 
import csv
import pandas as pd
```

```python
dados = []
path = '../data_raw/stores_sales_forecasting.csv'

with open(path, 'r', encoding='latin-1') as file:
    reader = csv.DictReader(file, delimiter=',')
    for row in reader:
        dados.append(row)
```

```python
dados[1]
```

```python
dados[0].keys()
```

## Ajustando nome das colunas

```python
key_mapping = {
    'Row ID': 'ID Registro',
    'Order ID': 'ID Pedido',
    'Order Date': 'Data do Pedido',
    'Ship Date': 'Data de Envio Pedido',
    'Ship Mode': 'Modo de Envio Pedido',
    'Customer ID': 'ID Cliente',
    'Customer Name': 'Nome Cliente',
    'Segment': 'Classificacao Cliente',
    'Country': 'Pais Venda',
    'City': 'Cidade Venda',
    'State': 'Estado Venda',
    'Postal Code': 'Codigo Postal Venda',
    'Region': 'Regiao Venda',
    'Product ID': 'ID Produto',
    'Category': 'Categoria Produto',
    'Sub-Category': 'Sub-Categoria Produto',
    'Product Name': 'Nome Produto',
    'Sales': 'Venda Total',
    'Quantity': 'Quantidade Vendida',
    'Discount': 'Desconto Aplicado',
    'Profit': 'Lucro Obtido'
}

novos_dados = []
for antigo in dados:
    dicionario_temporario = {}
    for chave_antiga, valor in antigo.items():
        dicionario_temporario[key_mapping[chave_antiga]] = valor
    novos_dados.append(dicionario_temporario)
```

```python
# Visualizando dados com chaves renomeadas
novos_dados[0]
```

## Transformando lista em dataframe

```python
ecommerce_df = pd.DataFrame(novos_dados)
ecommerce_df.head()
```

```python
ecommerce_df.info()
```

Nas colunas 'Venda Total', 'Desconto Aplicado' e 'Lucro Obtido' será necessário a transformação do tipo para float, e 'Quantidade Vendida' para inteiro. <br>
Também será necessário ajustar os tipos das Datas de Pedido e Envio

```python
# Verificando se há valores nulos
ecommerce_df.isnull().sum()
```

## Ajustando Tipos de Dados

```python
# Ajustando tipo de dados da coluna 'Venda Total', 'Desconto Aplicado' e 'Lucro Obtido' para flutuante
ecommerce_df['Venda Total'] = ecommerce_df['Venda Total'].astype(float)
ecommerce_df['Desconto Aplicado'] = ecommerce_df['Desconto Aplicado'].astype(float)
ecommerce_df['Lucro Obtido'] = ecommerce_df['Lucro Obtido'].astype(float)
```

```python
# Ajustando tipo de dados da coluna 'Quantidade Vendida' para inteiro
ecommerce_df['Quantidade Vendida'] = ecommerce_df['Quantidade Vendida'].astype(int)
```

```python
# Ajustando tipo de dados da coluna 'Data do Pedido' e 'Data de Envio Pedido' para datetime
ecommerce_df['Data do Pedido'] = pd.to_datetime(ecommerce_df['Data do Pedido'], format='%m/%d/%Y')
ecommerce_df['Data de Envio Pedido'] = pd.to_datetime(ecommerce_df['Data de Envio Pedido'], format='%m/%d/%Y')
```
